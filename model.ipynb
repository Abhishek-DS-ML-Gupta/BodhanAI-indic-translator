```python
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(
    repo_id="Bonenk/indic-translate-GGUF",
    filename="indic-translate-Q4_K_M.gguf"
)

print(model_path)
```

```python
from llama_cpp import Llama

llm = Llama(
    model_path=model_path,
    n_gpu_layers=-1,
    n_ctx=4096,
    n_batch=512,
    verbose=False
)

response = llm.create_completion(
    prompt="Translate the following text into Hindi: India is developing rapidly in artificial intelligence.",
    temperature=0,
    max_tokens=256
)

print(response["choices"][0]["text"].strip())
```

```python
response = llm.create_chat_completion(
    messages=[
        {
            "role": "user",
            "content": "Translate into Hindi: India is developing rapidly in artificial intelligence."
        }
    ],
    temperature=0,
    max_tokens=256
)

print(response["choices"][0]["message"]["content"])
```
